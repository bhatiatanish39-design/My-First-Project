```python
df = df.drop(['EmployeeCount', 'Over18', 'StandardHours'], axis=1)
```

```python
df.shape
```

```python
df['IncomeGroup'] = pd.cut(
    df['MonthlyIncome'],
    bins=[0, 3000, 6000, 10000, np.inf],
    labels=['Low', 'Medium', 'High', 'Very High']
)
```

```python
df
```

```python
df['AgeGroup'] = pd.cut(
    df['Age'],
    bins=[0, 25, 35, 45, 60],
    labels=['18-25', '26-35', '36-45', '46+']
)
```

```python
df
```

```python
df['Attrition'].value_counts()
```

```python
sns.countplot(data=df, x='Attrition')
plt.title('Employee Attrition Distribution')
plt.show()
```

```python
sns.countplot(data=df, x='Gender', hue='Attrition')
plt.title('Attrition by Gender')
plt.show()
```

```python
sns.boxplot(data=df, x='Attrition', y='Age')
plt.title('Age Distribution by Attrition')
plt.show()
```

```python
sns.countplot(data=df, x='MaritalStatus', hue='Attrition')
plt.title('Attrition by Marital Status')
plt.show()
```

```python
sns.countplot(
    data=df,
    x='JobRole',
    hue='Attrition'
)

plt.xticks(rotation=45)
plt.title('Attrition by Job Role')
plt.show()
```

```python
sns.boxplot(
    data=df,
    x='Attrition',
    y='MonthlyIncome'
)

plt.title('Monthly Income vs Attrition')
plt.show()
```

```python
sns.countplot(
    data=df,
    x='IncomeGroup',
    hue='Attrition'
)

plt.title('Attrition by Income Group')
plt.show()
```

```python
sns.countplot(
    data=df,
    x='JobSatisfaction',
    hue='Attrition'
)

plt.title('Job Satisfaction vs Attrition')
plt.show()
```

```python
sns.countplot(
    data=df,
    x='OverTime',
    hue='Attrition'
)

plt.title('Overtime vs Employee Attrition')
plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(15,10))
sns.heatmap(
    numeric_df.corr(),
    cmap='coolwarm',
    center=0
)

plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x='Age',
    y='MonthlyIncome',
    hue='Attrition'
)

plt.title('Age vs Monthly Income by Attrition')
plt.show()
```

# Key Business Insights

### 1. Overtime
Employees working overtime show a different attrition pattern compared
with employees who do not work overtime.

### 2. Compensation
Monthly income and income groups show differences between employees who
left and those who stayed.

### 3. Career Growth
Variables such as years since promotion and years in current role can
help identify employees with potentially higher attrition risk.

### 4. Job Role
Attrition varies across different job roles, indicating that retention
strategies may need to be role-specific.

### 5. Employee Satisfaction
Job and workplace satisfaction are important variables to examine when
understanding employee turnover.

# Business Recommendations

Based on the analysis, organizations could:

- Monitor employees with consistently high overtime.
- Improve career progression and promotion opportunities.
- Review compensation structures for high-risk employee groups.
- Develop role-specific retention strategies.
- Regularly monitor employee satisfaction.
- Identify employees showing multiple attrition-related characteristics.
