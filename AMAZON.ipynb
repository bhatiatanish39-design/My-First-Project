```python
top_10_revenue = (
    df.groupby("product_id")["total_revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

print(top_10_revenue)
```

```python
top_10_quantity = (
    df.groupby("product_id")["quantity_sold"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

print(top_10_quantity)
```

```python
revenue_by_category = df.groupby("product_category")["total_revenue"].sum().sort_values()

plt.figure(figsize=(8,5))
plt.bar(revenue_by_category.index, revenue_by_category.values)
plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
quantity_by_category = df.groupby("product_category")["quantity_sold"].sum().sort_values()

plt.figure(figsize=(8,5))
plt.bar(quantity_by_category.index, quantity_by_category.values)
plt.title("Quantity Sold by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Quantity Sold")
plt.xticks(rotation=45)
plt.show()
```

```python
top10 = df.groupby("product_id")["total_revenue"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
plt.barh(top10.index, top10.values)
plt.title("Top 10 Products by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Product ID")
plt.gca().invert_yaxis()
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["rating"], bins=10)
plt.title("Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.show()
```

```python
correlation = df.corr(numeric_only=True)

plt.figure(figsize=(8,6))
sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["discount_percent"], df["total_revenue"])
plt.title("Discount vs Revenue")
plt.xlabel("Discount (%)")
plt.ylabel("Revenue")
plt.show()
```

```python
revenue_by_category = df.groupby("product_category")["total_revenue"].sum()

plt.figure(figsize=(8,8))
plt.pie(
    revenue_by_category,
    labels=revenue_by_category.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Revenue Share by Product Category")
plt.axis("equal")
plt.show()
```

```python
region_revenue = df.groupby("customer_region")["total_revenue"].sum()

plt.figure(figsize=(8,8))
plt.pie(
    region_revenue,
    labels=region_revenue.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Revenue Share by Customer Region")
plt.axis("equal")
plt.show()
```
